```python
import numpy as np
import pandas as pd
import random

dataset = [
    {
        "intent": "greeting",
        "patterns": ["hi", "hello", "hey", "good morning", "good evening"],
        "responses": ["Hello! How can I help you?", "Hi there! What can I do for you?"]
    },
    {
        "intent": "goodbye",
        "patterns": ["bye", "see you", "good night", "talk to you later"],
        "responses": ["Goodbye!", "See you soon!"]
    },
    {
        "intent": "thanks",
        "patterns": ["thank you", "thanks", "thanks a lot"],
        "responses": ["You're welcome!", "Glad to help!"]
    },
    {
        "intent": "creator",
        "patterns": ["who created you?", "who made you?"],
        "responses": ["I was created by Selva Praveen’s chatbot project."]
    },
    {
        "intent": "weather",
        "patterns": ["what is the weather today?", "tell me today's weather", "weather report"],
        "responses": ["Please tell me your city for weather details."]
    },
    {
        "intent": "name",
        "patterns": ["what is your name?", "who are you?"],
        "responses": ["I am your chatbot assistant."]
    },
    {
        "intent": "age",
        "patterns": ["how old are you?", "what is your age?"],
        "responses": ["I don't have an age. I was created recently!"]
    },
        {
        "intent": "work",
        "patterns": ["what are you doing?"],
        "responses": ["Now I'm assisting you"]
    }
]
```

```python
pattern = []
response_map = {}
intent_pattern = {}

for item in dataset:
    intent_pattern[item['intent']] = item['responses']
    for p in item['patterns']:
        pattern.append(p)
        response_map[p] = item['responses']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

vector = TfidfVectorizer()
x = vector.fit_transform(pattern)
```

```python
def chat_response(user_input):
   user_vec = vector.transform([user_input])
   similarity = cosine_similarity(user_vec,x)

   idx = np.argmax(similarity)
   best_score = similarity[0][idx]

   if best_score < 0.5:
      return "Sorry I can't understand you"

   match_pattern = pattern[idx]
   return random.choice(response_map[match_pattern])
```

```python
print("Chatbot is ready to talk! Type 'exit' to quit.")
while True:
    msg = input("You : ")
    if msg.lower() == 'exit':
        print("Chatbot : Goodbye!")
        break
    print(f"You : {msg}")
    print("Chatbot :",chat_response(msg))
```

```python
import pickle

chatbot_data = {
    "pattern": pattern,
    "response_map": response_map,
    "vector": vector,
    "x": x,
}

with open("chatbot_data.pkl", "wb") as f:
    pickle.dump(chatbot_data, f)

print("Chatbot data saved to chatbot_data.pkl")
```
